In this example, we want to use ntrfc-methods to define a naca-profile.
We are using the naca_airfoil_creator module from ntrfc.turbo.airfoil_generators and d the DomainParameters object.

Lets start with importing all the necessary modules

```python
from ntrfc.turbo.airfoil_generators.naca_airfoil_creator import naca
import numpy as np
import pyvista as pv
from ntrfc.cascade_case.domain.geoparas import DomainParameters

pv.set_jupyter_backend("static")
```

We now can define 2D points using the naca-function.
Afterwards, we can create a PolyData object to parse this to the generate_params_by_pointcloud method to generate all attributes of DomainParas

```python
pv.set_jupyter_backend("static")
xs,ys = naca("6510",256)
points = pv.PolyData(np.stack([xs,ys,np.zeros(len(xs))]).T)
alpha = 1
domainparas = DomainParameters()
domainparas.generate_params_by_pointcloud(points,alpha=1)
domainparas.plot()
```

The domainaras object now has several parameters of the blade-geometry as attributes.
