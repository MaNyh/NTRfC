Often, one only get a point cloud as the description of an airfoil.
In this case we have to make sure to sort the points in a counterclockwise manner.
With the sorted profile points, one might extract the leading- and trailing-edge indices and derive other parameters.

```python
from ntrfc.cascade_case.domain import DomainParameters
import importlib.resources
import numpy as np
import pyvista as pv
import os

profilepoints_file = importlib.resources.files("ntrfc") / "data/turbine_cascade/profilepoints.txt"


points = np.loadtxt(profilepoints_file)
pointcloud = pv.PolyData(points)
turbine_domain = DomainParameters()
turbine_domain.generate_params_by_pointcloud(pointcloud)
```

```python
pv.set_jupyter_backend("static")

if os.getenv('DISPLAY') is None:
    pv.start_xvfb()  # Start X virtual framebuffer (Xvfb)

plotpath = turbine_domain.plot_domainparas()
```
