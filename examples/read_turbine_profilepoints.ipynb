```python
from ntrfc.turbo.pointcloud_methods import calc_concavehull
import pkg_resources
import numpy as np


profilepoints_file = pkg_resources.resource_filename("ntrfc","../data/turbine_cascade/profilepoints.txt")

points = np.loadtxt(profilepoints_file)[:, :2]
xs,ys = calc_concavehull(points[::,0],points[::,1])
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.plot(xs,ys)
plt.show()
```
