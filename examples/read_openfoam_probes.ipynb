```python
from fluidfoam.readpostpro import readprobes
import os
import importlib.resources

from ntrfc.turbo.cascade_case.solution import GenericCascadeCase


casedir =  importlib.resources.files("ntrfc") / "data/openfoam_probedata"

case = GenericCascadeCase(casedir)
variables = os.listdir(case.case_meta.case_root_directory / "postProcessing/probes/0")

probes_loc, time, u = readprobes(str(case.case_meta.case_root_directory), time_name = 'mergeTime', name = "U")
_, _, rho = readprobes(str(case.case_meta.case_root_directory), time_name = 'mergeTime', name = "rho")
_, _, p = readprobes(str(case.case_meta.case_root_directory), time_name = 'mergeTime', name = "p")
_, _, T = readprobes(str(case.case_meta.case_root_directory), time_name = 'mergeTime', name = "T")

valuedicts = []
for i in range(len(probes_loc)):
    valuedicts.append({"U":u[:,i], "rho":rho[:,i], "p":p[:,i], "T":T[:,i]})
    case.probes.add_probe(probes_loc[i], valuedicts[i],"line",time)
```

```python
from IPython.display import Image
import tempfile

tmpf = tempfile.NamedTemporaryFile(suffix='.png', delete=False)
case.probes.plot_probes(path=tmpf.name,value="U")
Image(tmpf.name)
```
