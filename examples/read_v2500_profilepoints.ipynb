```python
from ntrfc.cascade_case.domain import DomainParameters
import importlib.resources
import numpy as np
import pyvista as pv
import os


profilepoints_file = importlib.resources.files("ntrfc") / "data/v2500_compressorcascade/profilepoints.txt"


points = np.loadtxt(profilepoints_file)
pointcloud = pv.PolyData(points)
compressor_domain = DomainParameters()
compressor_domain.generate_params_by_pointcloud(pointcloud)
```

```python
pv.set_jupyter_backend("static")

if os.getenv('DISPLAY') is None:
    pv.start_xvfb()  # Start X virtual framebuffer (Xvfb)

plotpath = compressor_domain.plot_domainparas()
```

```python
from IPython.display import Image
Image(plotpath, width=400, height=400)
```
