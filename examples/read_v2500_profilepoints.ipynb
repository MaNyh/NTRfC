```python
from ntrfc.cascade_case.domain import DomainParameters
import pkg_resources
import numpy as np
import pyvista as pv


profilepoints_file = pkg_resources.resource_filename("ntrfc","../data/v2500_compressorcascade/profilepoints.txt")


points = np.loadtxt(profilepoints_file)
pointcloud = pv.PolyData(points)
compressor_domain = DomainParameters()
compressor_domain.generate_params_by_pointcloud(pointcloud)
```

```python
pv.set_jupyter_backend("static")

plotpath = compressor_domain.plot_domainparas()
```
