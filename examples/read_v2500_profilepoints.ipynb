```python
from ntrfc.geometry.alphashape import auto_concaveHull
import pkg_resources
import numpy as np


profilepoints_file = pkg_resources.resource_filename("ntrfc","../data/v2500_compressorcascade/profilepoints.txt")


points = np.loadtxt(profilepoints_file)[:, :2]
xs,ys, alpha = auto_concaveHull(points[::,0],points[::,1])
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.plot(xs,ys)
plt.show()
```
