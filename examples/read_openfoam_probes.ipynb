```python
from fluidfoam.readpostpro import readprobes
import os

from ntrfc.cascade_case.solution import GenericCascadeCase

case = GenericCascadeCase("../data/openfoam_probedata")
variables = os.listdir(case.case_meta.case_root_directory+"/postProcessing/probes/0")

probes_loc, time, u = readprobes("../data/openfoam_probedata", time_name = 'mergeTime', name = "U")
_, _, rho = readprobes("../data/openfoam_probedata", time_name = 'mergeTime', name = "rho")
_, _, p = readprobes("../data/openfoam_probedata", time_name = 'mergeTime', name = "p")
_, _, T = readprobes("../data/openfoam_probedata", time_name = 'mergeTime', name = "T")

valuedicts = []
for i in range(len(probes_loc)):
    valuedicts.append({"U":u[:,i], "rho":rho[:,i], "p":p[:,i], "T":T[:,i]})
    case.probes.add_probe(probes_loc[i], valuedicts[i],"line",time)
```
