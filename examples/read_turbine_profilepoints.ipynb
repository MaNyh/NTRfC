Often, one only get a point cloud as the description of an airfoil.
In this case we have to make sure to sort the points in a counterclockwise manner.
With the sorted profile points, one might extract the leading- and trailing-edge indices and derive other parameters.

```python
from ntrfc.turbo.pointcloud_methods import calc_concavehull
import pkg_resources
import numpy as np


profilepoints_file = pkg_resources.resource_filename("ntrfc","../data/turbine_cascade/profilepoints.txt")

points = np.loadtxt(profilepoints_file)[:, :2]
xs,ys = calc_concavehull(points[::,0],points[::,1],0.01)
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.plot(xs,ys)
plt.ylim(-0.1,0.1)
plt.xlim(-0.1,0.1)
plt.show()
```
